Langkah berikutnya, pilih satu halaman web yang akan di-scrap. Di sini saya menggunakan contoh sebuah halaman artikel berita dari detik.com. Periksa terlebih dahulu halaman web tersebut, contoh di sini mengambil elemen judul berita atau headline. Periksa terlebih dahulu dengan membuka halaman web yang akan dituju, lalu klik kanan pada judul berita/headline, dan pilih inspect element. Hasil yang diperoleh pada contoh adalah judul berupa elemen h1, dan class nya adalah detail__title.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL of the news page
url = 'https://news.detik.com/internasional/d-8382337/iran-klaim-hancurkan-radar-militer-terbesar-as-di-qatar'

# Send a request to the website
response = requests.get(url)

# Parse the HTML content
soup = BeautifulSoup(response.text, 'html.parser')

# Find news headlines (replace 'h2' and 'class' with findings from Opera Inspect)
# halaman web yang lain bahkan dari web utama yang sama, bisa jadi tag heading dan class-nya berbeda
headlines = soup.find_all('h1', class_='detail__title')

# Extract text and store in a list
news_list = []
for headline in headlines:
    news_list.append(headline.text.strip())

# Display the results
print(news_list)
# Or save to a DataFrame
df = pd.DataFrame(news_list, columns=['Headline'])
print(df)
```

### Jawaban Nomor 2: Mengambil Keseluruhan Teks Artikel

```python
# JAWABAN NOMOR 2: Mengambil Keseluruhan Teks pada Artikel Berita

import requests
from bs4 import BeautifulSoup

url = 'https://news.detik.com/internasional/d-8382337/iran-klaim-hancurkan-radar-militer-terbesar-as-di-qatar'

response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# LANGKAH-LANGKAH MENGAMBIL KESELURUHAN TEKS ARTIKEL:
# 1. Buka halaman web di browser
# 2. Klik kanan pada badan artikel, pilih "Inspect"
# 3. Cari tag HTML yang membungkus seluruh konten artikel
# 4. Pada detik.com, artikel biasanya berada dalam elemen <div class="detail__body-text">

# Cara 1: Menggunakan class tertentu (jika diketahui)
article_body = soup.find('div', class_='detail__body-text')

if article_body:
    # Ambil semua teks, hapus whitespace berlebih
    full_text = article_body.get_text(separator='\n', strip=True)
    print("=== CARA 1: Menggunakan class detail__body-text ===")
    print(full_text[:500])  # Tampilkan 500 karakter pertama
    print("...")
    print(f"\nTotal karakter: {len(full_text)}")
else:
    print("Elemen tidak ditemukan dengan class tersebut")
```

```python
# Cara 2: Mengambil semua paragraf <p> di dalam artikel
# Cara ini lebih umum dan bisa digunakan di berbagai website

paragraphs = soup.find_all('p')

# Filter paragraf yang memiliki konten (bukan script atau iklan)
article_paragraphs = []
for p in paragraphs:
    text = p.get_text(strip=True)
    # Filter paragraf yang terlalu pendek (biasanya iklan/navigasi)
    if len(text) > 50:
        article_paragraphs.append(text)

print("=== CARA 2: Mengambil semua paragraf <p> ===")
print(f"Ditemukan {len(article_paragraphs)} paragraf\n")

for i, para in enumerate(article_paragraphs[:5], 1):  # Tampilkan 5 paragraf pertama
    print(f"{i}. {para[:150]}...")
    print()

full_article = '\n\n'.join(article_paragraphs)
print(f"Total karakter artikel: {len(full_article)}")
```

### Jawaban Nomor 3: Web Scrapping Halaman Web Lain

**Website yang akan di-scrap:** https://www.kompas.com/
**Target:** Berita utama (headline) dari halaman depan Kompas.com

```python
# JAWABAN NOMOR 3: Web Scrapping Halaman Web Lain

import requests
from bs4 import BeautifulSoup
import pandas as pd

# Website target: Kompas.com
url = 'https://www.kompas.com/'

# Tambahkan headers untuk menghindari blocking (beberapa website memblokir bot)
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# LANGKAH MENEMUKAN ELEMEN YANG AKAN DI-SCRAPE:
# 1. Buka https://www.kompas.com/ di browser
# 2. Klik kanan pada judul berita utama, pilih "Inspect"
# 3. Perhatikan struktur HTML:
#    - Berita utama biasanya dalam <div class="most__wrap">
#    - Judul berada dalam tag <a class="article__link">
# 4. Catat tag dan class tersebut untuk digunakan di kode

# Mengambil judul berita utama
headlines = soup.find_all('a', class_='article__link', limit=10)

# Ekstrak teks dan link
news_data = []
for headline in headlines:
    title = headline.get_text(strip=True)
    link = headline.get('href', '')
    if title:  # Pastikan judul tidak kosong
        news_data.append({
            'Judul': title,
            'Link': link
        })

# Tampilkan hasil
df = pd.DataFrame(news_data)
print("=== HASIL WEB SCRAPING KOMPAS.COM ===")
print(f"Total berita ditemukan: {len(news_data)}\n")
print(df)

# Simpan ke CSV (opsional)
# df.to_csv('kompas_headlines.csv', index=False)
```

### Ringkasan Elemen yang Harus Dituliskan untuk Web Scraping

| Langkah | Keterangan |
|---------|------------|
| 1. Inspect Element | Buka website → Klik kanan pada elemen yang ingin diambil → Pilih "Inspect" |
| 2. Identifikasi Tag | Perhatikan tag HTML (h1, div, a, p, span, dll) |
| 3. Identifikasi Class/ID | Catat nilai class atau id dari elemen tersebut |
| 4. Gunakan di Kode | `soup.find('tag', class_='nama_class')` atau `soup.find_all('tag', class_='nama_class')` |

**Tips Penting:**
- Setiap website memiliki struktur HTML yang berbeda
- Selalu gunakan `headers` dengan `User-Agent` untuk menghindari blocking
- Beberapa website menggunakan JavaScript untuk memuat konten (memerlukan library tambahan seperti Selenium)
- Hormati `robots.txt` dan terms of service website yang di-scrape
